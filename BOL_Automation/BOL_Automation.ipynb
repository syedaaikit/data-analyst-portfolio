```python
import pdfplumber
import pandas as pd
import os

pdf_filename = "Sample_BOL.pdf"
csv_output = "extracted_shipments.csv"

def extract_bol_data(pdf_path):
    extracted_records = []

    with pdfplumber.open(pdf_path) as pdf:
        for page_num, page in enumerate(pdf.pages, start=1):
            text = page.extract_text()

            record = {
                "Document_Name": pdf_filename,
                "Page_Number": page_num,
                "Extracted_Content_Snippet": text[:200].replace("\n", " ") if text else "No text found",
                "Processing_Status": "Successfully Extracted"
            }
            extracted_records.append(record)

    return extracted_records

if os.path.exists(pdf_filename):
    data = extract_bol_data(pdf_filename)

    if data:
        # Convert data into a Pandas DataFrame
        df = pd.DataFrame(data)

        # Save to CSV
        df.to_csv(csv_output, index=False, encoding="utf-8")

        print("Pipeline completed successfully! Here is the preview of extracted data (head):\n")

        # Using the head() function to display the top rows
        print(df.head())
    else:
        print("Warning: No text extracted from the PDF.")
else:
    print(f"Error: File '{pdf_filename}' not found.")
```
